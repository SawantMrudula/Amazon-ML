```python
import pandas as pd
import os
import requests
from PIL import Image
import pytesseract
import re
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
import xgboost as xgb
from sklearn.metrics import mean_squared_error
import numpy as np

# Function to download images
def download_images(image_link, save_path):
    response = requests.get(image_link)
    if response.status_code == 200:
        # Extract filename from image_link
        filename = image_link.split('/')[-1]
        # Create complete save path using filename
        complete_save_path = os.path.join(save_path, filename)
        with open(complete_save_path, 'wb') as f:
            f.write(response.content)
    else:
        print(f"Failed to download image from {image_link}")

# Function to extract text using pytesseract with improved preprocessing
def extract_text_from_image_pytesseract(image_path):
    img = Image.open(image_path)
    extracted_text = pytesseract.image_to_string(img)
    extracted_text = re.sub(r'[^\d\.\s\w]', '', extracted_text)  # Remove non-numeric characters
    return extracted_text

# Function to split numeric values
# Function to split numeric values
def split_value(entity_value):
    pattern = r"(\d*\.?\d+)\s*([a-zA-Z]+)"
    match = re.match(pattern, entity_value)
    if match:
        value, _ = match.groups()
        return float(value)
    return 0.0 # Return 0.0 if no match is found to avoid None values

# Function to split units
def split_value_unit(entity_value):
    pattern = r"(\d*\.?\d+)\s*([a-zA-Z]+)"
    match = re.match(pattern, entity_value)
    if match:
        _, unit = match.groups()
        return unit.lower()
    return None

# Normalizing function for allowed units
def get_allowed_units(train_data):
    pattern = r"(\d*\.?\d+)\s*([a-zA-Z]+)"
    units = []
    for entity_value in train_data['entity_value']:
        match = re.search(pattern, entity_value)
        if match:
            _, unit = match.groups()
            units.append(unit.lower())
    return list(set(units))

# Function to normalize text
def normalize_text(extracted_text, allowed_units):
    pattern = r"(\d*\.?\d+)\s*([a-zA-Z]+)"
    match = re.search(pattern, extracted_text)
    if match:
        value, unit = match.groups()
        unit = unit.lower()
        if unit in allowed_units:
            return f"{float(value)} {unit}"
    return extracted_text

# Feature extraction
def extract_features(data, allowed_units, is_train=True):
    features = []
    labels_values = []
    labels_unit = []

    for idx, row in data.iterrows():
        image_path = f"/content/images/{row['image_link'].split('/')[-1]}" # Extract filename from image_link

        if not os.path.exists(image_path):
            download_images(row['image_link'], '/content/images')

        extracted_text = extract_text_from_image_pytesseract(image_path)
        extracted_text = normalize_text(extracted_text, allowed_units)
        features.append({
            'group_id': row['group_id'],
            'entity_name': row['entity_name'],
            'extracted_text': extracted_text
        })

        if is_train:
            labels_values.append(split_value(row['entity_value']))
            labels_unit.append(split_value_unit(row['entity_value']))

    if is_train:
        return pd.DataFrame(features), labels_values
    else:
        return pd.DataFrame(features), None

# Train model with XGBoost
def train_model(X_train, y_train):
    vectorizer = TfidfVectorizer()
    X_train_combined = X_train.apply(lambda row: f"{row['group_id']} {row['entity_name']} {row['extracted_text']}", axis=1)
    X_train_vectors = vectorizer.fit_transform(X_train_combined)

    dtrain = xgb.DMatrix(X_train_vectors, label=y_train)
    params = {
        'objective': 'reg:squarederror',
        'eval_metric': 'rmse',
        'learning_rate': 0.05,
        'max_depth': 8,
        'alpha': 5,
        'n_estimators': 300
    }
    model = xgb.train(params, dtrain, num_boost_round=150)
    return model, vectorizer

# Prediction function
def predict(model, vectorizer, test_data):
    test_features = test_data.apply(lambda row: f"{row['group_id']} {row['entity_name']} {row['extracted_text']}", axis=1)
    test_vectors = vectorizer.transform(test_features)
    dtest = xgb.DMatrix(test_vectors)
    predictions = model.predict(dtest)
    return predictions

# Create output file with an additional column for actual value and error
def create_output_file(predictions, test_data):
    actual_values = test_data['entity_value'].apply(split_value)
    output = pd.DataFrame({
        'image_link': test_data['image_link'],
        'group_id': test_data['group_id'],
        'prediction': predictions,
        'actual_value': actual_values,
        'error': np.abs(actual_values - predictions)  # Calculate the error
    })
    output.to_csv('/content/test_out.csv', index=False)
    return output

# Main pipeline
def main():
    train_data = pd.read_csv('/content/train.csv')

    allowed_units = get_allowed_units(train_data)
    X_train, y_train_values = extract_features(train_data, allowed_units, is_train=True)

    # Train and validation split
    X_train_split, X_val_split, y_train_split, y_val_split = train_test_split(X_train, y_train_values, test_size=0.2, random_state=42)

    model_values, vectorizer_values = train_model(X_train_split, y_train_split)

    # Validation
    y_pred_values = predict(model_values, vectorizer_values, X_val_split)
    val_error_values = mean_squared_error(y_val_split, y_pred_values)
    print(f"Validation MSE: {val_error_values}")

    # Test predictions
    test_data = pd.read_csv('/content/train.csv')
    test_features, _ = extract_features(test_data, allowed_units, is_train=False)
    test_predictions_values = predict(model_values, vectorizer_values, test_features)

    # Create output file with errors and actual values
    output = create_output_file(test_predictions_values, test_data)
    print(output.head())

if __name__ == "__main__":
    main()
```

```python
# prompt: calculate accuracy formula using prediction,actual_value and error

def calculate_accuracy(predictions, actual_values, error_threshold):
  """
  Calculates the accuracy of predictions based on an error threshold.

  Args:
    predictions: A list or array of predicted values.
    actual_values: A list or array of actual values.
    error_threshold: The maximum allowed error for a prediction to be considered accurate.

  Returns:
    The accuracy as a percentage.
  """
  correct_predictions = sum(abs(pred - actual) <= error_threshold for pred, actual in zip(predictions, actual_values))
  total_predictions = len(predictions)
  accuracy = (correct_predictions / total_predictions) * 100
  return accuracy
```
